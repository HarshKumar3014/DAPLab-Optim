```python
# ===========================================================================
#  AdamW vs Muon: diagnosing optimization during fine-tuning on SST-2
#
#  Premise: the two optimizers differ in GEOMETRY, not just tuning. AdamW
#  normalizes coordinate-wise (update ~ sign(g), spectrally concentrated).
#  Muon orthogonalizes the momentum matrix -- steepest descent under the
#  spectral norm, so every singular value of the update is ~1. That is
#  falsifiable, so we measure it directly instead of only comparing accuracy,
#  which saturates on SST-2.
#
#  Run: paste into one Colab cell on a T4. ~3h. Completed runs are cached.
# ===========================================================================
!pip install -q transformers datasets

import json, math, random, time
from collections import defaultdict
from dataclasses import dataclass, asdict, replace
from pathlib import Path

import numpy as np, torch, torch.nn.functional as F
import matplotlib; matplotlib.use("Agg"); import matplotlib.pyplot as plt
from torch.utils.data import DataLoader
from torch.nn.attention import sdpa_kernel, SDPBackend
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from datasets import load_dataset
from IPython.display import display, Markdown, Image

DEV = torch.device("cuda")
# T4 is Turing: no bf16 tensor cores. fp16 autocast + GradScaler for training;
# fp32 for Newton-Schulz; fp32 with autocast off for all curvature probes.
BF16 = torch.cuda.is_bf16_supported()
AMP, NS_DT = (torch.bfloat16, torch.bfloat16) if BF16 else (torch.float16, torch.float32)
RES = Path("results"); (RES / "figures").mkdir(parents=True, exist_ok=True)
OPTS = ("adamw", "muon"); COLOR = {"adamw": "#1f77b4", "muon": "#d62728"}
LR_GRID = [1e-5, 3e-5, 1e-4, 3e-4, 1e-3, 3e-3]

@dataclass
class Cfg:
    model: str = "HuggingFaceTB/SmolLM2-135M"
    optimizer: str = "adamw"; lr: float = 3e-5; aux_lr: float = 3e-4
    momentum: float = 0.9    # beta1 for AdamW, mu for Muon; 0 = ablation
    wd: float = 0.01; steps: int = 400; bs: int = 32; seed: int = 0
    probe_n: int = 128; probe_bs: int = 8; full_probe: bool = True
    tag: str = ""

# =============================================================== Muon =====
@torch.no_grad()
def newton_schulz(G, steps=5):
    """Orthogonalize G -> ~UV^T by a quintic Newton-Schulz iteration.

    The coefficients do not converge to exact UV^T; they saturate fast for
    singular values in (0,1], leaving a spectrum of ~[0.7, 1.2] after 5 steps.
    The Frobenius pre-normalization is MANDATORY: the iteration only converges
    for spectral norm <= 1, and ||G||_2 <= ||G||_F guarantees that.
    """
    a, b, c = 3.4445, -4.7750, 2.0315
    X = G.to(NS_DT)
    T = X.size(-2) > X.size(-1)
    if T: X = X.mT                                   # cheaper when rows <= cols
    X = X / (X.norm(dim=(-2, -1), keepdim=True) + 1e-7)
    for _ in range(steps):
        A = X @ X.mT
        X = a * X + (b * A + c * (A @ A)) @ X
    return (X.mT if T else X).to(G.dtype)

class Muon(torch.optim.Optimizer):
    """Muon for >=2D hidden weights only. Embeddings, head, biases and norm
    gains go to AdamW -- they are not feature-to-feature maps."""
    def __init__(self, params, lr, momentum=0.95, weight_decay=0.0):
        super().__init__(params, dict(lr=lr, momentum=momentum, weight_decay=weight_decay))

    @torch.no_grad()
    def step(self):
        for g_ in self.param_groups:
            mom, lr, wd = g_["momentum"], g_["lr"], g_["weight_decay"]
            for p in g_["params"]:
                if p.grad is None or not torch.isfinite(p.grad).all():
                    continue                          # fp16 overflow -> scaler rescales
                g = p.grad
                if mom > 0:
                    st = self.state[p]
                    buf = st.setdefault("buf", torch.zeros_like(g))
                    buf.lerp_(g, 1 - mom)             # buf = mom*buf + (1-mom)*g
                    u = g.lerp(buf, mom)              # Nesterov lookahead
                else:
                    u = g                             # momentum-free ablation
                # 0.2*sqrt(max(m,n)) matches the update's per-entry RMS to
                # AdamW's, which is what makes a shared LR grid meaningful.
                scale = 0.2 * math.sqrt(max(p.shape[-2], p.shape[-1]))
                if wd: p.mul_(1 - lr * wd)            # decoupled
                p.add_(newton_schulz(u), alpha=-lr * scale)

def split_params(model):
    """The SAME split is used in both arms, so the AdamW baseline is not
    secretly a different experiment."""
    skip = ("embed", "lm_head", "score", "classifier", "head", "pooler")
    mats, aux = [], []
    for n, p in model.named_parameters():
        (mats if p.ndim >= 2 and not any(s in n.lower() for s in skip) else aux).append((n, p))
    return mats, aux

# ========================================================= diagnostics =====
@torch.no_grad()
def stable_rank(A):
    """||A||_F^2/||A||_2^2. 1 = rank-one, min(m,n) = isotropic. The most direct
    test of the AdamW-vs-Muon story."""
    f = A.pow(2).sum()
    return 0.0 if f == 0 else float(f / torch.linalg.matrix_norm(A, ord=2) ** 2)

@torch.no_grad()
def spectrum_entropy(A):
    """Normalized entropy of the singular values. 1.0 = flat spectrum (what an
    orthogonalized update looks like), ->0 = all energy in one direction."""
    s = torch.linalg.svdvals(A); p = s / s.sum(); p = p[p > 0]
    return float(-(p * p.log()).sum() / math.log(len(p)))

@torch.no_grad()
def geometry(W0_ref, named):
    out = {}
    for n, W0 in W0_ref.items():
        W = named[n].detach().float().cpu(); dW = W - W0
        out[n] = {"stable_rank_dW": stable_rank(dW),
                  "spectrum_entropy_dW": spectrum_entropy(dW),
                  "rel_update_norm": float(dW.norm() / W0.norm()),
                  "W_norm": float(W.norm())}   # for the reparameterization check
    return out

# Three sharpness estimators, because they disagree -- and the disagreement is
# itself the answer to "how reliable are these conclusions?".
# allow_unused/materialize_grads: real models have unused params (tied weights).
def _grad(lf, bs, ps, graph=False):
    tot, n = None, 0
    for b in bs:
        k = b["labels"].numel()
        g = torch.autograd.grad(lf(b) * k, ps, create_graph=graph,
                                allow_unused=True, materialize_grads=True)
        tot = list(g) if tot is None else [x + y for x, y in zip(tot, g)]; n += k
    return [x / n for x in tot]

def _hvp(lf, bs, ps, v):
    hv, n = [torch.zeros_like(p) for p in ps], 0
    for b in bs:
        k = b["labels"].numel()
        g = torch.autograd.grad(lf(b) * k, ps, create_graph=True,
                                allow_unused=True, materialize_grads=True)
        h = torch.autograd.grad(sum((a * c).sum() for a, c in zip(g, v)), ps,
                                allow_unused=True, materialize_grads=True)
        hv = [x + y.detach() for x, y in zip(hv, h)]; n += k
        del g, h
    return [x / n for x in hv]

_dot = lambda a, b: sum(float((x * y).sum()) for x, y in zip(a, b))
_nrm = lambda a: math.sqrt(sum(float(x.pow(2).sum()) for x in a))

def top_eig(lf, bs, ps, iters=20):
    """lambda_max by power iteration on HVPs. NOT reparameterization invariant."""
    v = [torch.randn_like(p) for p in ps]; n = _nrm(v); v = [x / n for x in v]
    lam = prev = None
    for _ in range(iters):
        hv = _hvp(lf, bs, ps, v)
        lam = _dot(hv, v)                                   # Rayleigh quotient
        n = _nrm(hv) + 1e-12; v = [x / n for x in hv]
        if prev and abs(lam - prev) / abs(prev) < 1e-3: break
        prev = lam
    return lam

def trace(lf, bs, ps, n_probes=15):
    """Hutchinson. Average curvature, not worst-case: lambda_max can be
    dominated by one stiff direction the optimizer never travels in. The std
    matters -- a difference inside it is not a difference."""
    e = [_dot(_hvp(lf, bs, ps, v), v) for v in
         ([torch.randint_like(p, high=2) * 2.0 - 1.0 for p in ps] for _ in range(n_probes))]
    t = torch.tensor(e); return float(t.mean()), float(t.std())

@torch.no_grad()
def _filter_dir(ps):
    """Random direction, per-row-normalized to the weight row (Li et al. 2018).
    Without it, a network with larger weights looks flatter for free."""
    ds = []
    for p in ps:
        if p.ndim < 2: ds.append(torch.zeros_like(p)); continue
        d = torch.randn_like(p)
        pf, df = p.view(p.size(0), -1), d.view(d.size(0), -1)
        ds.append((df * (pf.norm(dim=1, keepdim=True) /
                         (df.norm(dim=1, keepdim=True) + 1e-10))).view_as(p))
    return ds

@torch.no_grad()
def _loss(lf, bs):
    return sum(float(lf(b)) * b["labels"].numel() for b in bs) / sum(b["labels"].numel() for b in bs)

def profile(lf, bs, ps, alphas, n_dirs=3):
    torch.manual_seed(0)
    base, curves = [p.detach().clone() for p in ps], []
    try:
        for _ in range(n_dirs):
            d = _filter_dir(base); row = []
            for a in alphas:
                with torch.no_grad():
                    for p, b0, di in zip(ps, base, d): p.copy_(b0 + a * di)
                row.append(_loss(lf, bs))
            curves.append(row)
    finally:                                   # never leave the model perturbed
        with torch.no_grad():
            for p, b0 in zip(ps, base): p.copy_(b0)
    c = torch.tensor(curves); return c.mean(0).tolist(), c.std(0).tolist()

def adaptive_sharpness(lf, bs, ps, rho=0.05, steps=5):
    """max_{||eps/|theta|||<=rho} L(theta+eps) - L(theta), by projected gradient
    ascent in the rescaled space T=diag(|theta|).

    Dinh et al. (2017): any minimum can be made arbitrarily sharp by layer-wise
    rescaling, so raw lambda_max between optimizers ending at different ||W|| is
    confounded. This is the invariant estimator to quote for a head-to-head
    claim (Kwon et al. 2021)."""
    base = [p.detach().clone() for p in ps]; sc = [b.abs() + 1e-8 for b in base]
    resc = lambda xs: math.sqrt(sum(float((x / s).pow(2).sum()) for x, s in zip(xs, sc)))
    try:
        L0 = _loss(lf, bs); eps = [torch.zeros_like(b) for b in base]
        for _ in range(steps):
            with torch.no_grad():
                for p, b0, e in zip(ps, base, eps): p.copy_(b0 + e)
            gt = [g * (s ** 2) for g, s in zip(_grad(lf, bs, ps), sc)]  # ascent in T-space
            eps = [e + 0.25 * rho * x / (resc(gt) + 1e-12) for e, x in zip(eps, gt)]
            if (n := resc(eps)) > rho: eps = [e * (rho / n) for e in eps]  # project
        with torch.no_grad():
            for p, b0, e in zip(ps, base, eps): p.copy_(b0 + e)
        return _loss(lf, bs) - L0
    finally:
        with torch.no_grad():
            for p, b0 in zip(ps, base): p.copy_(b0)

# ========================================================= data + train ====
_CACHE = {}
def get_data(tok):
    """SST-2 test labels are hidden -> hold 2k out of TRAIN as dev for all model
    selection, report the official validation split (872) once as test. OOD =
    rotten_tomatoes: same task, longer reviews. SST-2 accuracy saturates, so the
    generalization gap is the more informative signal."""
    if "d" in _CACHE: return _CACHE["d"]
    s = load_dataset("glue", "sst2"); tr = s["train"].shuffle(seed=1234)
    f = lambda ds, k: ds.map(lambda b: dict(tok(b[k], truncation=True, max_length=128),
                                            labels=b["label"]),
                             batched=True, remove_columns=ds.column_names)
    _CACHE["d"] = (f(tr.select(range(2000, len(tr))), "sentence"), f(tr.select(range(2000)), "sentence"),
                   f(s["validation"], "sentence"), f(load_dataset("rotten_tomatoes")["test"], "text"))
    return _CACHE["d"]

def loader(ds, tok, bs, shuffle=False, seed=0):
    def coll(fs):
        y = torch.tensor([f.pop("labels") for f in fs])
        return dict(tok.pad(fs, return_tensors="pt"), labels=y)
    g = torch.Generator(); g.manual_seed(seed)
    return DataLoader(ds, batch_size=bs, shuffle=shuffle, collate_fn=coll,
                      generator=g if shuffle else None, drop_last=shuffle)

@torch.no_grad()
def evaluate(model, dl):
    model.eval(); c = n = 0
    for b in dl:
        b = {k: v.to(DEV) for k, v in b.items()}; y = b.pop("labels")
        with torch.autocast("cuda", dtype=AMP):
            c += int((model(**b).logits.float().argmax(-1) == y).sum())
        n += y.numel()
    model.train(); return c / n

def probe(cfg, model, probe_dl, W0_ref):
    """Curvature with dropout off, autocast off, and the MATH attention backend:
    the flash/mem-efficient SDPA kernels have no double-backward, so
    create_graph=True fails on them."""
    model.eval()
    out = {"geometry": geometry(W0_ref, dict(model.named_parameters()))}
    if cfg.full_probe:
        def lf(b):
            b = {k: v.to(DEV) for k, v in b.items()}; y = b.pop("labels")
            return F.cross_entropy(model(**b).logits.float(), y)
        ps, al = [p for p in model.parameters()], np.linspace(-.5, .5, 11).tolist()
        with sdpa_kernel(SDPBackend.MATH):
            out["lambda_max"] = top_eig(lf, probe_dl, ps)
            out["hessian_trace"], out["hessian_trace_std"] = trace(lf, probe_dl, ps)
            out["adaptive_sharpness"] = adaptive_sharpness(lf, probe_dl, ps)
            m, s = profile(lf, probe_dl, ps, al)
        out["profile"] = {"alphas": al, "mean": m, "std": s}
    model.train(); return out

def run(cfg):
    """One run -> one JSON. Cached runs are reused; a cached run whose probe
    failed is recomputed rather than silently keeping the hole."""
    path = RES / f"{cfg.tag}.json"
    if path.exists():
        log = json.load(open(path))
        if "probe_error" not in (log["probes"][-1] if log["probes"] else {}):
            print(f"[{cfg.tag}] cached"); return log

    random.seed(cfg.seed); np.random.seed(cfg.seed); torch.manual_seed(cfg.seed)
    tok = AutoTokenizer.from_pretrained(cfg.model)
    tok.pad_token = tok.pad_token or tok.eos_token
    model = AutoModelForSequenceClassification.from_pretrained(
        cfg.model, num_labels=2, dtype=torch.float32).to(DEV)
    model.config.pad_token_id = tok.pad_token_id

    train, dev, test, ood = get_data(tok)
    tl = loader(train, tok, cfg.bs, True, cfg.seed)
    dl, sl, ol = (loader(d, tok, 64) for d in (dev, test, ood))
    # Fixed probe set: curvature must be measured on identical data in both arms.
    probe_dl = list(loader(dev.select(range(cfg.probe_n)), tok, cfg.probe_bs))

    mats, aux = split_params(model)
    mp = [p for _, p in mats]
    opt_m = (Muon(mp, cfg.lr, cfg.momentum, cfg.wd) if cfg.optimizer == "muon" else
             torch.optim.AdamW(mp, cfg.lr, (cfg.momentum, 0.95), weight_decay=cfg.wd))
    # Aux group is AdamW at a FIXED lr in both arms, so the sweep isolates the
    # matrix optimizer. This is the fairness contract.
    opt_a = torch.optim.AdamW([p for _, p in aux], cfg.aux_lr, (0.9, 0.95))
    scaler = torch.amp.GradScaler("cuda", enabled=not BF16)

    idx = np.linspace(0, len(mats) - 1, 8).astype(int)
    W0_ref = {mats[i][0]: mats[i][1].detach().float().cpu().clone() for i in idx}

    log = {"cfg": asdict(cfg), "steps": [], "evals": [], "probes": [],
           "tracked_order": [mats[i][0] for i in idx]}
    warm, cos_hist, prev_g = int(.06 * cfg.steps), [], None
    model.train(); it = iter(tl); t0 = time.time()

    for step in range(1, cfg.steps + 1):
        s = step / warm if step < warm else \
            .5 * (1 + math.cos(math.pi * (step - warm) / (cfg.steps - warm)))
        for o, base in ((opt_m, cfg.lr), (opt_a, cfg.aux_lr)):
            for g in o.param_groups: g["lr"] = base * s

        try: b = next(it)
        except StopIteration: it = iter(tl); b = next(it)
        b = {k: v.to(DEV) for k, v in b.items()}; y = b.pop("labels")
        opt_m.zero_grad(True); opt_a.zero_grad(True)
        with torch.autocast("cuda", dtype=AMP):
            loss = F.cross_entropy(model(**b).logits.float(), y)
        scaler.scale(loss).backward()
        # fp16: unscale BOTH optimizers before clipping, or you clip a
        # 65536x-scaled gradient and the threshold means nothing.
        scaler.unscale_(opt_m); scaler.unscale_(opt_a)
        gn = float(torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0))

        # cos(g_t, g_{t-1}): momentum buys variance reduction, and it buys less
        # when consecutive gradients are already aligned.
        g = torch.cat([p.grad.flatten() for p in mp if p.grad is not None]).float()
        cos = float(F.cosine_similarity(g, prev_g, dim=0)) if prev_g is not None else None
        if torch.isfinite(g).all(): prev_g = g; cos_hist += [cos] if cos else []
        scaler.step(opt_m); scaler.step(opt_a); scaler.update()

        if step % 10 == 0 or step == 1:
            log["steps"].append({"step": step, "train_loss": float(loss.detach()), "grad_norm": gn,
                                 "lr": cfg.lr * s, "grad_cos": cos})
        if step % 100 == 0:
            log["evals"].append({"step": step, "dev_acc": evaluate(model, dl)})

    log["probes"].append(dict(probe(cfg, model, probe_dl, W0_ref), step=cfg.steps))
    log["wall_clock_s"] = time.time() - t0
    log["final"] = {k: {"acc": evaluate(model, d)}
                    for k, d in (("dev", dl), ("test", sl), ("ood", ol))}
    log["grad_cos_mean"] = float(np.mean(cos_hist)) if cos_hist else None
    json.dump(log, open(path, "w"), indent=1)
    print(f"[{cfg.tag}] {log['wall_clock_s']:.0f}s test={log['final']['test']['acc']:.4f} "
          f"ood={log['final']['ood']['acc']:.4f}")
    del model, opt_m, opt_a; torch.cuda.empty_cache()
    return log

# ============================================================= self-test ===
# Estimators checked against a closed form before they are trusted on a real
# model. ~5s. The SDPA check is the exact failure mode that silently emptied
# the sharpness columns on the first attempt.
def self_test():
    G = torch.randn(64, 128) @ torch.diag(torch.logspace(0, -3, 128))
    O = newton_schulz(G.cuda()).float().cpu()
    print(f"[NS]   stable rank {stable_rank(G):.2f} -> {stable_rank(O):.2f} | "
          f"entropy {spectrum_entropy(G):.3f} -> {spectrum_entropy(O):.3f}")
    A = torch.randn(12, 12); H = (A.T @ A + torch.eye(12)).cuda()
    w = torch.nn.Parameter(torch.randn(12).cuda())
    lf, bs = lambda b: .5 * w @ (H @ w), [{"labels": torch.zeros(1)}]
    lam, (tr, sd) = top_eig(lf, bs, [w], 60), trace(lf, bs, [w], 40)
    print(f"[curv] lambda_max {lam:.2f} vs exact {float(torch.linalg.eigvalsh(H).max()):.2f} | "
          f"trace {tr:.1f}±{sd:.1f} vs exact {float(H.trace()):.1f}")
    q = torch.randn(2, 2, 8, 16, requires_grad=True, device=DEV)
    with sdpa_kernel(SDPBackend.MATH):
        g = torch.autograd.grad(F.scaled_dot_product_attention(q, q, q).pow(2).sum(),
                                [q], create_graph=True)
        torch.autograd.grad((g[0] * torch.randn_like(q)).sum(), [q])
    print("[attn] double-backward through SDPA math backend: OK\n")
self_test()

# ============================================================ experiments ==
tag = lambda k, c: {"sweep": f"sweep_{c.optimizer}_lr{c.lr:g}",
                    "main": f"main_{c.optimizer}_s{c.seed}",
                    "mom": f"mom_{c.optimizer}_m{c.momentum:g}_s{c.seed}"}[k]

def load(pre):
    out = []
    for f in sorted(RES.glob(f"{pre}*.json")):
        r = json.load(open(f)); r["_tag"] = f.stem; out.append(r)
    return out

def best_lr(o):
    """By DEV accuracy only. Test is reported once, at the end."""
    c = [r for r in load("sweep") if r["cfg"]["optimizer"] == o]
    return max(c, key=lambda r: r["final"]["dev"]["acc"])["cfg"]["lr"]

# Stage 1 -- LR sweep. The SHAPE is a result: a wide stable basin is a real
# practical advantage, independent of peak accuracy. Each arm gets its own
# sweep and we compare best-vs-best; one shared LR would just measure who
# happens to like that LR.
for o in OPTS:
    for lr in LR_GRID:
        c = Cfg(optimizer=o, lr=lr, steps=200, full_probe=False)
        run(replace(c, tag=tag("sweep", c)))

# Stage 2 -- main runs, 3 seeds, full curvature probes.
for o in OPTS:
    lr = best_lr(o); print(f"\n{o} @ lr={lr:g}")
    for s in range(3):
        c = Cfg(optimizer=o, lr=lr, seed=s)
        run(replace(c, tag=tag("main", c)))

# Stage 3 -- momentum ablation. The project's hypothesis is that momentum
# matters less in post-training than in pretraining. Read alongside the
# measured gradient alignment. Note Newton-Schulz is scale-invariant (it
# Frobenius-normalizes its input), so for Muon this is a PURE direction
# ablation with no effective-LR confound.
for o in OPTS:
    lr = best_lr(o)
    for m in ((0.9 if o == "adamw" else 0.95), 0.0):
        for s in range(4):
            c = Cfg(optimizer=o, lr=lr, momentum=m, seed=s, full_probe=False)
            run(replace(c, tag=tag("mom", c)))

# ================================================== figures and tables =====
def ms(xs):
    a = np.array([x for x in xs if x is not None], float)
    return (a.mean(), a.std(ddof=1) if a.size > 1 else 0.0) if a.size else (None, None)

def show(fig, name):
    fig.tight_layout(); fig.savefig(RES / "figures" / name, dpi=150); plt.close(fig)
    display(Image(filename=str(RES / "figures" / name)))

fig, ax = plt.subplots(figsize=(5.4, 3.4))
for o in OPTS:
    p = sorted((r["cfg"]["lr"], r["final"]["dev"]["acc"])
               for r in load("sweep") if r["cfg"]["optimizer"] == o)
    ax.plot(*zip(*p), "o-", color=COLOR[o], label=o)
ax.set_xscale("log"); ax.set(xlabel="learning rate (matrix group)",
                             ylabel="dev accuracy", title="LR sensitivity")
ax.grid(alpha=.3); ax.legend(); show(fig, "lr_sweep.png")

fig, ax = plt.subplots(1, 3, figsize=(13, 3.5))
for o in OPTS:
    rs = [r for r in load("main") if r["cfg"]["optimizer"] == o]
    st = [s["step"] for s in rs[0]["steps"]]
    L = np.array([[s["train_loss"] for s in r["steps"]] for r in rs])
    sm = np.array([np.convolve(l, np.ones(10) / 10, "valid") for l in L])
    ax[0].plot(st[9:], sm.mean(0), color=COLOR[o], label=o)
    ax[0].fill_between(st[9:], sm.mean(0) - sm.std(0), sm.mean(0) + sm.std(0),
                       color=COLOR[o], alpha=.2)
    A = np.array([[e["dev_acc"] for e in r["evals"]] for r in rs])
    ax[1].errorbar([e["step"] for e in rs[0]["evals"]], A.mean(0), A.std(0),
                   color=COLOR[o], marker="o", capsize=3, label=o)
    C = np.array([[s["grad_cos"] for s in r["steps"] if s["grad_cos"]] for r in rs])
    ax[2].plot(st[-C.shape[1]:], C.mean(0), color=COLOR[o], alpha=.85, label=o)
for a, t in zip(ax, ["train loss", "dev accuracy", r"$\cos(g_t, g_{t-1})$"]):
    a.set(xlabel="step", title=t); a.grid(alpha=.3); a.legend()
show(fig, "training_curves.png")

# The core mechanistic figure.
fig, ax = plt.subplots(1, 3, figsize=(13.5, 3.6))
keys = [("stable_rank_dW", "stable rank of $\\Delta W$"),
        ("rel_update_norm", "$\\|\\Delta W\\|_F/\\|W_0\\|_F$"),
        ("spectrum_entropy_dW", "spectral entropy of $\\Delta W$")]
for o in OPTS:
    G = [r["probes"][-1]["geometry"] for r in load("main") if r["cfg"]["optimizer"] == o]
    names = list(G[0])
    for a, (k, lab) in zip(ax, keys):
        V = np.array([[g[n][k] for n in names] for g in G])
        a.errorbar(np.arange(len(names)), V.mean(0), V.std(0), marker="o",
                   color=COLOR[o], capsize=3, label=o)
        a.set_ylabel(lab)
for a in ax:
    a.set_xlabel("tracked matrix (shallow $\\rightarrow$ deep)"); a.grid(alpha=.3); a.legend()
ax[0].set_yscale("log"); show(fig, "geometry.png")

fig, ax = plt.subplots(figsize=(5.4, 3.4))
for o in OPTS:
    P = [r["probes"][-1]["profile"] for r in load("main") if r["cfg"]["optimizer"] == o]
    Y = np.array([p["mean"] for p in P])
    ax.errorbar(P[0]["alphas"], Y.mean(0), Y.std(0), marker="o", color=COLOR[o],
                capsize=3, label=o)
ax.set(xlabel=r"$\alpha$ along filter-normalized directions", ylabel="loss",
       title="1D loss profile around the solution")
ax.grid(alpha=.3); ax.legend(); show(fig, "sharpness_profile.png")

cols = ["optimizer", "lr", "dev", "test", "OOD", "$\\lambda_{max}$", "tr(H)",
        "adaptive sharp.", "grad cos", "\\|W\\|", "s/step"]
rows = ["| " + " | ".join(cols) + " |", "|" + "---|" * len(cols)]
for o in OPTS:
    rs = [r for r in load("main") if r["cfg"]["optimizer"] == o]
    pr = [r["probes"][-1] for r in rs]
    f = lambda v, fmt="{:.3f}": (lambda m, s: "-" if m is None else
                                 f"{fmt.format(m)} ± {fmt.format(s)}")(*ms(v))
    rows.append("| " + " | ".join([o, f"{rs[0]['cfg']['lr']:g}"] + [
        f([r["final"][k]["acc"] for r in rs]) for k in ("dev", "test", "ood")] + [
        f([p["lambda_max"] for p in pr], "{:.2f}"),
        f([p["hessian_trace"] for p in pr], "{:.1f}"),
        f([p["adaptive_sharpness"] for p in pr], "{:.4f}"),
        f([r["grad_cos_mean"] for r in rs]),
        f([sum(g["W_norm"] for g in p["geometry"].values()) for p in pr], "{:.1f}"),
        f([r["wall_clock_s"] / r["cfg"]["steps"] for r in rs], "{:.3f}")]) + " |")

by = defaultdict(list)
for r in load("mom"):
    by[(r["cfg"]["optimizer"], r["cfg"]["momentum"])].append(r["final"]["dev"]["acc"])
mrows = ["| optimizer | momentum | dev acc | n | Δ vs. momentum-free |", "|---|---|---|---|---|"]
for o in OPTS:
    off, _ = ms(by[(o, 0.0)])
    for (oo, m), v in sorted(by.items()):
        if oo == o:
            mu, sd = ms(v)
            mrows.append(f"| {o} | {m:g} | {mu:.3f} ± {sd:.3f} | {len(v)} | "
                         f"{'—' if m == 0 else f'{mu-off:+.3f}'} |")

srows = ["| optimizer | lr | dev acc |", "|---|---|---|"] + [
    f"| {r['cfg']['optimizer']} | {r['cfg']['lr']:g} | {r['final']['dev']['acc']:.3f} |"
    for r in sorted(load("sweep"), key=lambda r: (r["cfg"]["optimizer"], r["cfg"]["lr"]))]

summary = ("## Results\n\n### Main comparison (mean ± std over 3 seeds)\n\n" + "\n".join(rows)
           + "\n\n### Momentum ablation\n\n" + "\n".join(mrows)
           + "\n\n### LR sweep\n\n" + "\n".join(srows) + "\n")
(RES / "summary.md").write_text(summary); display(Markdown(summary))
```
